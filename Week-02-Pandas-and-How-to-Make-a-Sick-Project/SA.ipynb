### Clean the Data

Well, we can’t get our answers with what we currently have, so you’ll have to make some changes.

Let’s not worry about anything below the first 5 values for now. Convert the top 5 to a single canonical value―say, `'US'`, which is nice and short.

```python
df_salary_tech["country"] = df_salary_tech["country"].replace({
    "United States": "US",
    "USA": "US",
    "U.S.": "US",
    "United States of America": "US",
})
```

Have a look at the count of each unique country again now.

```python
df_salary_tech["country"].value_counts().head(10)
```

Did you notice anything interesting?

```python
import re

def normalize_country(value):
    if pd.isna(value):
        return value

    value = str(value).strip()
    normalized = value.lower()

    if re.fullmatch(r"u\.?s\.?a\.?|u\.?s\.?", normalized):
        return "US"
    if "united states" in normalized or "usa" in normalized:
        return "US"
    if "canada" in normalized:
        return "Canada"
    if "united kingdom" in normalized or "uk" in normalized or "england" in normalized:
        return "United Kingdom"
    return value


df_salary_tech["country"] = df_salary_tech["country"].apply(normalize_country)
```

```python
df_salary_tech["country"].value_counts().head(10)
```

It’s looking good so far. Let’s find out the minimum, mean, and maximum (in that order) salary by state, sorted by the mean in descending order.

```python
df_salary_tech["salary"] = pd.to_numeric(
    df_salary_tech["salary"].astype(str).str.replace(",", "", regex=False),
    errors="coerce",
)

salary_by_state = (
    df_salary_tech[df_salary_tech["country"] == "US"]
    .groupby("state")["salary"]
    .agg(["min", "mean", "max"])
    .sort_values("mean", ascending=False)
)

salary_by_state
```

### Bonus #0

Apply the same steps as before to `df_salary`, but do not filter for any specific industry. Do perform the other data cleaning stuff, and get to a point where you can generate the minimum, mean, and maximum by state.

```python
df_salary_filtered = df_salary.copy()

df_salary_filtered["country"] = df_salary_filtered["country"].replace({
    "United States": "US",
    "USA": "US",
    "U.S.": "US",
    "United States of America": "US",
})

df_salary_filtered = df_salary_filtered[
    (df_salary_filtered["currency"] == "USD")
    & (df_salary_filtered["country"] == "US")
].copy()

df_salary_filtered["salary"] = pd.to_numeric(
    df_salary_filtered["salary"].astype(str).str.replace(",", "", regex=False),
    errors="coerce",
)

df_salary_filtered = df_salary_filtered.dropna(subset=["state", "salary"])

salary_by_state_all = (
    df_salary_filtered.groupby("state")["salary"]
    .agg(["min", "mean", "max"])
    .sort_values("mean", ascending=False)
)

salary_by_state_all
```

### Bonus #2

Filter out the non-single-states (e.g., `'California, Colorado'`) in the most elegant way possible (i.e., *not* by blacklisting all the bad values).

```python
single_state_salary = df_salary_filtered[
    ~df_salary_filtered["state"].str.contains(",", na=False)
]

single_state_salary.groupby("state")["salary"].agg(["min", "mean", "max"]).sort_values("mean", ascending=False)
```

### Bonus #3

Show the quantiles instead of just minimum, mean, and maximum―say 0%, 5%, 25%, 50%, 75%, 95%, and 100%. Outliers may be deceiving.

Sort by whatever interests you―like say the *50th* percentile.

And throw in a count by state too. It would be interesting to know how many data points contribute to the figures for each state. (*Hint: your nice formatting from Bonus #1 might not work this time around.* 😜)

```python
quantile_levels = [0, 0.05, 0.25, 0.5, 0.75, 0.95, 1]

single_state_salary = df_salary_filtered[
    ~df_salary_filtered["state"].str.contains(",", na=False)
].copy()

salary_by_state_quantiles = (
    single_state_salary.groupby("state")["salary"]
    .quantile(quantile_levels)
    .unstack()
    .rename(columns=lambda col: f"p{col:g}")
    .assign(count=single_state_salary.groupby("state")["salary"].count())
    .sort_values("p0.5", ascending=False)
)

salary_by_state_quantiles
```
